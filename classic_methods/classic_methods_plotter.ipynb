# PULSAR: Classic Methods Benchmark Visualizer & Comparison Suite
This notebook follows the exact visualization conventions used across the PULSAR repository (`dataset_plotter.ipynb` and the `TimesFM 3.0` evaluation plots) so you can inspect both **`fixed`** and **`varying`** radar profile benchmarks:
1. **Formatted Benchmark Tables** (`RMSE`, `MAE`, `MedAE`, `MAPE`, `Gate Hit Rate`, `Latency`)
2. **Teammate-Style Single-Method Forecast Plots** (`Observed Context`, `Ground Truth`, `Model Forecast`, `±10 μs Tactical Gate`, `Prediction Error Gap`)
3. **3-Method Overlay Plots & Bar Charts** (`Rolling Median` vs. `Period Estimation` vs. `Hidden Markov Model`)

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Image, Markdown

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'classic_methods' else Path.cwd()
RESULTS_ROOT = PROJECT_ROOT / 'results' / 'classic_methods'

for mode in ['fixed', 'varying']:
    comp_csv = RESULTS_ROOT / f'comparison_{mode}_pri.csv'
    if comp_csv.exists():
        df_comp = pd.read_csv(comp_csv)
        display(Markdown(f'### 1. Side-by-Side Metrics Table ({mode.capitalize()} Profile - PRI)'))
        display(df_comp)
```

```python
for mode in ['fixed', 'varying']:
    bar_chart = RESULTS_ROOT / 'plots_comparison' / f'{mode}_pri' / 'summary_metrics_bar_chart.png'
    if bar_chart.exists():
        display(Markdown(f'### 2. Summary Bar Chart ({mode.capitalize()} Profile - PRI)'))
        display(Image(filename=str(bar_chart)))
```

```python
# Select PROFILE_MODE ('fixed' or 'varying'), RADAR_TYPE ('stable', 'jitter', 'stagger', 'dwell_and_switch', 'sliding'),
# and ERROR_LEVEL ('clean', 'error_lv1', 'error_lv2', 'error_lv3') to inspect:
PROFILE_MODE = 'varying'
RADAR_TYPE = 'dwell_and_switch'
ERROR_LEVEL = 'error_lv1'

display(Markdown(f'### 3. Teammate-Style Forecast Plots: `{RADAR_TYPE}` ({ERROR_LEVEL}) [{PROFILE_MODE.capitalize()} Profile]'))
for method_slug in ['rolling_median', 'period_estimation', 'hidden_markov_model']:
    p_img = RESULTS_ROOT / method_slug / f'{PROFILE_MODE}_pri' / 'plots' / f'{RADAR_TYPE}_{ERROR_LEVEL}.png'
    if p_img.exists():
        display(Image(filename=str(p_img)))

display(Markdown(f'### 4. Combined 3-Method Overlay Plot: `{RADAR_TYPE}` ({ERROR_LEVEL}) [{PROFILE_MODE.capitalize()} Profile]'))
overlay_img = RESULTS_ROOT / 'plots_comparison' / f'{PROFILE_MODE}_pri' / f'{RADAR_TYPE}_{ERROR_LEVEL}.png'
if overlay_img.exists():
    display(Image(filename=str(overlay_img)))
```
