# PULSAR: Classic Methods Benchmark Visualizer & Comparison Suite
This notebook follows the exact visualization conventions used across the PULSAR repository (`dataset_plotter.ipynb` and the `TimesFM 3.0` evaluation plots) so you can inspect:
1. **Formatted Benchmark Tables** (`RMSE`, `MAE`, `MedAE`, `MAPE`, `Gate Hit Rate`, `Latency`)
2. **Teammate-Style Single-Method Forecast Plots** (`Observed Context`, `Ground Truth`, `Model Forecast`, `±10 μs Tactical Gate`, `Prediction Error Gap`)
3. **3-Method Overlay Plots & Bar Charts** (`Rolling Median` vs. `Period Estimation` vs. `Hidden Markov Model`)

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Image, Markdown

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'classic_methods' else Path.cwd()
RESULTS_ROOT = PROJECT_ROOT / 'results' / 'classic_methods'

comp_csv = RESULTS_ROOT / 'comparison_fixed_pri.csv'
df_comp = pd.read_csv(comp_csv)
display(Markdown('### 1. Side-by-Side Metrics Table (Fixed Profile - PRI)'))
display(df_comp)
```

```python
display(Markdown('### 2. Summary Bar Chart (RMSE & Tactical Gate Hit Rate)'))
bar_chart = RESULTS_ROOT / 'plots_comparison' / 'fixed_pri' / 'summary_metrics_bar_chart.png'
if bar_chart.exists():
    display(Image(filename=str(bar_chart)))
```

```python
# Select any radar_type ('stable', 'jitter', 'stagger', 'dwell_and_switch', 'sliding')
# and error_level ('clean', 'error_lv1', 'error_lv2', 'error_lv3') to inspect:
RADAR_TYPE = 'dwell_and_switch'
ERROR_LEVEL = 'error_lv1'

display(Markdown(f'### 3. Teammate-Style Forecast Plots: `{RADAR_TYPE}` ({ERROR_LEVEL})'))
for method_slug in ['rolling_median', 'period_estimation', 'hidden_markov_model']:
    p_img = RESULTS_ROOT / method_slug / 'fixed_pri' / 'plots' / f'{RADAR_TYPE}_{ERROR_LEVEL}.png'
    if p_img.exists():
        display(Image(filename=str(p_img)))

display(Markdown(f'### 4. Combined 3-Method Overlay Plot: `{RADAR_TYPE}` ({ERROR_LEVEL})'))
overlay_img = RESULTS_ROOT / 'plots_comparison' / 'fixed_pri' / f'{RADAR_TYPE}_{ERROR_LEVEL}.png'
if overlay_img.exists():
    display(Image(filename=str(overlay_img)))
```
