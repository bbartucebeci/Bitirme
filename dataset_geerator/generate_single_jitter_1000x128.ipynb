# Generate Single Gaussian Jitter Radar Dataset With Error Levels 1, 2, 3

This notebook generates synthetic PRI datasets for one Gaussian jitter radar.

The output is saved under `datasets/single_jitter_1000x128_<error_level>/`.

## Imports

```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
import yaml
```

# Error Level Definitions

```python
ERROR_LEVELS = {
    "clean": {
        "poi": 1.0,
        "multipath_prob": 0.0,
        "aoa_drift_rate": 0.0,
        "toa_noise_std": 0.0,
        "freq_noise_std": 0.0,
        "pw_noise_std": 0.0,
        "amp_noise_std": 0.0,
        "aoa_noise_std": 0.0,
        "global_drop_rate": 0.0,
        "noise_rate": 0.0,
        "quantize": False,
    },
    "error_lv1": {
        "poi": 0.995,
        "multipath_prob": 0.0,
        "aoa_drift_rate": 0.0,
        "toa_noise_std": 0.1,
        "freq_noise_std": 0.1,
        "pw_noise_std": 0.01,
        "amp_noise_std": 0.2,
        "aoa_noise_std": 0.05,
        "global_drop_rate": 0.0,
        "noise_rate": 1.0,
        "quantize": False,
    },
    "error_lv2": {
        "poi": 0.99,
        "multipath_prob": 0.01,
        "aoa_drift_rate": 0.2,
        "toa_noise_std": 0.5,
        "freq_noise_std": 0.5,
        "pw_noise_std": 0.03,
        "amp_noise_std": 0.5,
        "aoa_noise_std": 0.2,
        "global_drop_rate": 0.005,
        "noise_rate": 5.0,
        "quantize": False,
    },
    "error_lv3": {
        "poi": 0.98,
        "multipath_prob": 0.03,
        "aoa_drift_rate": 0.5,
        "toa_noise_std": 1.0,
        "freq_noise_std": 1.0,
        "pw_noise_std": 0.05,
        "amp_noise_std": 1.0,
        "aoa_noise_std": 0.5,
        "global_drop_rate": 0.01,
        "noise_rate": 10.0,
        "quantize": False,
    },
}
```

# Paths

```python
PROJECT_ROOT = Path("..").resolve()

RADAR_TYPE = "jitter"
SEQUENCE_COUNT = 1000
PULSES_PER_SEQUENCE = 128
RANDOM_SEED = 42

GENERATOR_DIR = PROJECT_ROOT / "synthetic data generation"
DATASET_ROOT = PROJECT_ROOT / "datasets"

sys.path.append(str(GENERATOR_DIR))
from radar_generator import Environment, JitterRadar

RADAR_CLASS = JitterRadar

PROJECT_ROOT, GENERATOR_DIR
```

## Dataset Configuration

This dataset contains one Gaussian jitter radar.

Its mean PRI is 1000 microseconds. The PRI varies naturally with a Gaussian
standard deviation of 50 microseconds. This intrinsic jitter is present even
in the clean dataset.

Frequency is constant at 9000 MHz. The clean level has no additional
measurement noise, missing pulses, false pulses, multipath, or AOA drift.
The other levels use the error settings above.

```python
RADAR_ID = "R_JITTER"
PRI_US = 1000.0
PRI_JITTER_STD_US = 50.0
PRI_ARGUMENTS = {
    "pri_mean": PRI_US,
    "pri_jitter_std": PRI_JITTER_STD_US,
}
PRI_METADATA = {
    "pri_mean_us": PRI_US,
    "pri_jitter_std_us": PRI_JITTER_STD_US,
}

FREQ_MEAN_MHZ = 9000.0
PW_MEAN_US = 1.5
AMP_MEAN_DBM = -55.0
AOA_MEAN_DEG = 45.0
```

# Save CSV Function Definition

```python
def save_individual_sequence_csvs(split_df, split_dir):
    for sequence_id, seq_df in split_df.groupby("sequence_id"):
        seq_df = seq_df.sort_values("pulse_index")
        seq_df.to_csv(split_dir / f"{sequence_id}.csv", index=False)
```

## Generate Pulse Sequences and Save Split CSV Files

Each `sequence_id` represents one independent radar pulse train.

Do not concatenate different sequences as one continuous signal. The train,
validation, and test splits contain different sequence IDs.

```python
for LEVEL, cfg in ERROR_LEVELS.items():
    DATASET_NAME = f"single_{RADAR_TYPE}_1000x128_{LEVEL}"
    DATASET_DIR = DATASET_ROOT / DATASET_NAME

    TRAIN_DIR = DATASET_DIR / "train"
    VAL_DIR = DATASET_DIR / "val"
    TEST_DIR = DATASET_DIR / "test"

    TRAIN_DIR.mkdir(parents=True, exist_ok=True)
    VAL_DIR.mkdir(parents=True, exist_ok=True)
    TEST_DIR.mkdir(parents=True, exist_ok=True)

    all_sequences = []

    for i in range(SEQUENCE_COUNT):
        sequence_id = f"seq_{i:04d}"
        env = Environment()

        radar = RADAR_CLASS(
            radar_id=RADAR_ID,
            **PRI_ARGUMENTS,
            freq_mean=FREQ_MEAN_MHZ,
            freq_mode="constant",
            freq_hopping_values=None,
            poi=cfg["poi"],
            multipath_prob=cfg["multipath_prob"],
            multipath_delay_mean=1.5,
            multipath_delay_std=0.2,
            pw_mean=PW_MEAN_US,
            amp_mean=AMP_MEAN_DBM,
            aoa_mean=AOA_MEAN_DEG,
            scan_period=0.0,
            aoa_drift_rate=cfg["aoa_drift_rate"],
            stretch_pw_on_multipath=False,
        )
        env.add_radar(radar)

        df_seq = env.generate(
            duration=PULSES_PER_SEQUENCE * PRI_US,
            toa_noise_std=cfg["toa_noise_std"],
            freq_noise_std=cfg["freq_noise_std"],
            pw_noise_std=cfg["pw_noise_std"],
            amp_noise_std=cfg["amp_noise_std"],
            aoa_noise_std=cfg["aoa_noise_std"],
            global_drop_rate=cfg["global_drop_rate"],
            noise_rate=cfg["noise_rate"],
            quantize=cfg["quantize"],
        )

        df_seq = df_seq.sort_values("TOA").reset_index(drop=True)
        df_seq["sequence_id"] = sequence_id
        df_seq["pulse_index"] = np.arange(len(df_seq))
        df_seq["radar_type"] = RADAR_TYPE
        df_seq["error_level"] = LEVEL

        all_sequences.append(df_seq)

    df = pd.concat(all_sequences, ignore_index=True)

    sequence_ids = sorted(df["sequence_id"].unique())
    rng = np.random.default_rng(RANDOM_SEED)
    rng.shuffle(sequence_ids)

    train_ids = set(sequence_ids[:700])
    val_ids = set(sequence_ids[700:800])
    test_ids = set(sequence_ids[800:1000])

    train_df = df[df["sequence_id"].isin(train_ids)].copy()
    val_df = df[df["sequence_id"].isin(val_ids)].copy()
    test_df = df[df["sequence_id"].isin(test_ids)].copy()

    train_df.to_csv(TRAIN_DIR / "sequences.csv", index=False)
    val_df.to_csv(VAL_DIR / "sequences.csv", index=False)
    test_df.to_csv(TEST_DIR / "sequences.csv", index=False)

    save_individual_sequence_csvs(train_df, TRAIN_DIR)
    save_individual_sequence_csvs(val_df, VAL_DIR)
    save_individual_sequence_csvs(test_df, TEST_DIR)

    metadata = {
        "name": DATASET_NAME,
        "radar_type": RADAR_TYPE,
        "sequence_count": SEQUENCE_COUNT,
        "expected_pulses_per_sequence": PULSES_PER_SEQUENCE,
        "error_level": LEVEL,
        "error_parameters": cfg,
        "files": {
            "train": "train/sequences.csv",
            "val": "val/sequences.csv",
            "test": "test/sequences.csv",
            "individual_sequence_files":
                "Each split folder also contains one CSV per sequence, named seq_XXXX.csv.",
        },
        "split": {
            "train": {
                "ratio": 0.70,
                "sequence_count": train_df["sequence_id"].nunique(),
                "row_count": len(train_df),
            },
            "val": {
                "ratio": 0.10,
                "sequence_count": val_df["sequence_id"].nunique(),
                "row_count": len(val_df),
            },
            "test": {
                "ratio": 0.20,
                "sequence_count": test_df["sequence_id"].nunique(),
                "row_count": len(test_df),
            },
        },
        "radar_parameters": {
            "radar_id": RADAR_ID,
            **PRI_METADATA,
            "freq_mean_mhz": FREQ_MEAN_MHZ,
            "freq_mode": "constant",
            "poi": cfg["poi"],
            "multipath_prob": cfg["multipath_prob"],
            "aoa_drift_rate": cfg["aoa_drift_rate"],
            "pw_mean_us": PW_MEAN_US,
            "amp_mean_dbm": AMP_MEAN_DBM,
            "aoa_mean_deg": AOA_MEAN_DEG,
        },
        "environment_parameters": {
            "duration_us": PULSES_PER_SEQUENCE * PRI_US,
            "toa_noise_std": cfg["toa_noise_std"],
            "freq_noise_std": cfg["freq_noise_std"],
            "pw_noise_std": cfg["pw_noise_std"],
            "amp_noise_std": cfg["amp_noise_std"],
            "aoa_noise_std": cfg["aoa_noise_std"],
            "global_drop_rate": cfg["global_drop_rate"],
            "noise_rate": cfg["noise_rate"],
            "quantize": cfg["quantize"],
        },
        "units": {
            "TOA": "microseconds",
            "PRI": "microseconds",
            "Frequency": "MHz",
            "PW": "microseconds",
            "Amplitude": "dBm",
            "AOA": "degrees",
        },
    }

    with open(DATASET_DIR / "metadata.yaml", "w", encoding="utf-8") as f:
        yaml.safe_dump(metadata, f, sort_keys=False)

    print(DATASET_NAME)
    print("Train:", train_df["sequence_id"].nunique(), "seqs,", len(train_df), "rows")
    print("Val:", val_df["sequence_id"].nunique(), "seqs,", len(val_df), "rows")
    print("Test:", test_df["sequence_id"].nunique(), "seqs,", len(test_df), "rows")
```

```python
readme_path = DATASET_ROOT / "README.md"

if readme_path.exists():
    print(f"Existing dataset README preserved: {readme_path}")
else:
    readme = """# PRI Benchmark Datasets

Each dataset is stored in a separate folder named:

single_<radar_type>_1000x128_<error_level>

Every dataset folder contains metadata.yaml and three splits:

- train: 700 independent sequences
- val: 100 independent sequences
- test: 200 independent sequences

Each split has a combined sequences.csv and individual seq_XXXX.csv files.
Rows belonging to different sequence_id values must not be treated as one
continuous pulse train.

The exact radar settings and error parameters are in metadata.yaml. The
1000x128 name denotes 1000 sequences and a generation duration based on
128 mean-PRI intervals. Missing, false, and multipath pulses can change
the number of saved rows.
"""
    DATASET_ROOT.mkdir(parents=True, exist_ok=True)
    readme_path.write_text(readme, encoding="utf-8")
    print(f"Created: {readme_path}")
```
